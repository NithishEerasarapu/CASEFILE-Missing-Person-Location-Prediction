# 6. Location Prediction

A Random Forest model is used to predict the geographical location from GPS movement data.

## Objective

The objective is to estimate a probable latitude and longitude based on historical movement patterns.

## Features Used

- User ID
- Hour
- Minute
- Day
- Month

## Method

1. Load the cleaned GPS dataset.
2. Convert User IDs into numerical values.
3. Extract time-based features.
4. Train the Random Forest model.
5. Predict latitude and longitude.
6. Evaluate the prediction errors.

The predicted location is an estimate based on the available data. It does not prove the actual location of a person.
