# 5. Anomaly Detection

Isolation Forest is used to identify unusual GPS movement patterns in the dataset.

## Objective

The goal is to detect GPS records that differ significantly from normal movement patterns.

## Features Used

- Latitude
- Longitude
- Distance
- Speed

## Method

1. Load the clustered GPS dataset.
2. Select movement-related features.
3. Apply the Isolation Forest algorithm.
4. Label records as Normal or Anomaly.
5. Save the results for further analysis.

An anomaly indicates an unusual movement pattern. It does not automatically indicate suspicious or criminal activity.
