# 7. Route Prediction

A Markov Chain is used to analyze possible movement transitions between different location clusters.

## Objective

The objective is to understand how movement changes from one geographical cluster to another.

## Method

1. Load the clustered GPS dataset.
2. Sort the records by user, date, and time.
3. Identify the previous location cluster.
4. Calculate transition probabilities between clusters.
5. Use these probabilities to analyze possible movement routes.

The transition probabilities represent patterns learned from the dataset. They are not a guarantee of a person's future route.
