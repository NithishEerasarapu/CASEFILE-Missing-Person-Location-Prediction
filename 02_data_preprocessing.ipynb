# 2. Data Preprocessing

Data preprocessing prepares the GPS dataset for machine learning.

The following steps are performed:

1. Load the GPS dataset.
2. Check for missing values.
3. Remove duplicate records.
4. Convert Date and Time into proper formats.
5. Check latitude and longitude values.
6. Extract useful time-based features such as hour, day, month, and weekday.
7. Save the cleaned dataset for further analysis.

The cleaned data is then used for feature engineering and machine learning models.
