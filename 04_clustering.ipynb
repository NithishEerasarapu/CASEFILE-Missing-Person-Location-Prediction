# 4. Movement Clustering

K-Means clustering is used to group GPS locations based on their geographical similarity.

## Objective

The goal is to identify different movement areas or location clusters from the GPS data.

## Method

1. Load the feature-engineered GPS dataset.
2. Select Latitude and Longitude.
3. Scale the selected features.
4. Apply K-Means clustering.
5. Create 5 location clusters.
6. Save the clustered data for further analysis.

Clustering helps identify frequently occurring geographical areas in the dataset.
